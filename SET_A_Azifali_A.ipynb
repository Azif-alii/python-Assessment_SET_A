Question 1 –
Electricity Bill Calculator

```python
customer_name = input("Enter customer name : ")
units_consumed = int(input("Enter units consumed : "))
if units_consumed < 0:
    print("Invalid units")
else:
    if units_consumed <= 100:
        electricity_charge = units_consumed * 2
    elif units_consumed <= 200:
        electricity_charge = (100 * 2) + ((units_consumed - 100) * 3)
    elif units_consumed <= 400:
        electricity_charge = (100 * 2) + (100 * 3) + ((units_consumed - 200) * 5)
    else:
        electricity_charge = (100 * 2) + (100 * 3) + (200 * 5) + ((units_consumed - 400) * 7)
    service_charge = 50
    total_bill = electricity_charge + service_charge
    print("Name of the Customer : ", customer_name)
    print("Units Consumed by the Customer :", units_consumed)
    print("Electricity Charge : ", electricity_charge)
    print("Service Charge : ", service_charge)
    print("Total Bill : ", total_bill)
```

Question 2 –
Student Result Management


```python
n = int(input("Enter number of students: "))
passed = 0
failed = 0
highest_average = 0
top_performer = " "
for i in range(n):
    name = input("Enter student name: ")
    python_marks = int(input("Enter Python marks: "))
    database_marks = int(input("Enter Database marks: "))
    aptitude_marks = int(input("Enter Aptitude marks: "))
    average = (python_marks + database_marks + aptitude_marks) / 3
    if python_marks >= 40 and database_marks >= 40 and aptitude_marks >= 40 and average >= 50:
        result = "Passed"
        if average >=80 and average <= 100:
            grade = "A"
        elif average >=70 and average <= 79:
            grade = "B"
        elif average >= 60 and average <= 69:
            grade = "C"
        else:
            grade = "D"
        print(name, "- Average:", average, "- Passed - Grade:", grade)
        passed += 1
        if average > highest_average:
            highest_average = average
            top_performer = name
    else:
        result = "Failed"
        print(name, "- Average:", average, "- Failed")
        failed += 1
print("Passed:", passed)
print("Failed:", failed)
print("Top Performer:", top_performer)
```

Question 3 –
Online Shopping Cart

```python
class Product:
    def __init__(self, id, name, price, stock):
        self.id = id
        self.name = name
        self.price = price
        self.stock = stock


class ShoppingCart:
    def __init__(self, products):
        self.products = products
        self.cart = {}

    def add(self, id, qty):
        if id not in self.products:
            print("Product not found")
        elif qty <= 0:
            return
        elif qty > self.products[id].stock:
            print("Insufficient stock")
        else:
            self.cart[id] = self.cart.get(id, 0) + qty
            self.products[id].stock -= qty

    def remove(self, id, qty):
        if id not in self.cart:
            print("Product not in cart")
        else:
            if qty >= self.cart[id]:
                qty = self.cart[id]

            self.cart[id] -= qty
            self.products[id].stock += qty

            if self.cart[id] == 0:
                del self.cart[id]

    def total(self):
        total = 0

        for id in self.cart:
            total += self.products[id].price * self.cart[id]

        return total

    def display(self):
        total = self.total()

        for id in self.cart:
            p = self.products[id]
            qty = self.cart[id]
            subtotal = p.price * qty

            print(p.name, "- Quantity:", qty, "- Subtotal:", subtotal)

        if total >= 20000:
            discount = 15
        elif total >= 10000:
            discount = 10
        elif total >= 5000:
            discount = 5
        else:
            discount = 0

        discount_amount = total * discount // 100
        final = total - discount_amount

        print("Total Amount:", total)
        print("Discount:", str(discount) + "%")
        print("Discount Amount:", discount_amount)
        print("Final Amount:", final)
```
